**Load joblib version**

```python
import pandas as pd
from joblib import load
```

```python
joblib_path="/content/sample_data/diabetes_model_pipeline.joblib"
```

```python
# Load the saved model

loaded_joblib_pipeline = load(joblib_path)
```

```python
new_data = pd.DataFrame({"Pregnancies": [2],
                         "Glucose":[120],
                         "BloodPressure":[70],
                         "SkinThickness":[25],
                         "Insulin":[80],
                         "BMI": [28.5],
                         "DiabetesPedigreeFunction":[0.45],
                         "Age":[35]})
```

```python
# data will be scaled and passed on to the model for prediction inside the pipeline
prediction = loaded_joblib_pipeline.predict(new_data)
print("Predicted class:", prediction[0])

if prediction[0] == 1:
    print("⚠️ Model predicts: Diabetes")
else:
    print("✅ Model predicts: No Diabetes")
```
